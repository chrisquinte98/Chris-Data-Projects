### Tarea 4: Modularización con Funciones

```python
def resumen_estadistico(numeros):
    
    suma_total = sum(numeros)
    cantidad = len(numeros)
    promedio = suma_total / cantidad

    return {
    "suma": suma_total,
    "promedio": promedio,
    "cantidad": cantidad
    }

# Ejemplo
resultado = resumen_estadistico([10, 20, 30])

print(f"Suma: {resultado['suma']}")
print(f"Promedio: {resultado['promedio']}")
print(f"Cantidad: {resultado['cantidad']}")
```
